```python
import spacy.cli
spacy.cli.download("en_core_web_lg")
nlp = spacy.load("en_core_web_lg")
```

```python
doc = nlp("dog cat banana adkgmnkng")

for token in doc:
  print(token.text, "Vector:", token.has_vector, "OOV: ", token.is_oov)
```

```python
doc[0].vector.shape
```

```python
base_token = nlp("bread")
base_token.vector.shape
```

```python
# similarity = mostly appear in the same context
doc = nlp("bread sandwich burger cat tiger human wheat man woman")

for token in doc:
  print(f"{token.text} <-> {base_token.text}: ", token.similarity(base_token))
```

```python
def print_similarity(base_word, word_to_compare):
  base_token = nlp(base_word)
  doc = nlp(word_to_compare)

  for token in doc:
    print(f"{token.text} <-> {base_token.text}: ", token.similarity(base_token))
```

```python
print_similarity('iphone', 'apple samsung iphone dog dog cat macbook galaxy ipad')
```

**When 2 words appear in the same context too often, it will have high similarity**

```python
king = nlp.vocab["king"].vector
man = nlp.vocab["man"].vector
woman = nlp.vocab["woman"].vector
queen = nlp.vocab["queen"].vector

result = king - man + woman
```

```python
from sklearn.metrics.pairwise import cosine_similarity

cosine_similarity([result], [queen])
```
