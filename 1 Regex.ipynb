# Use Regex for Chatbot Cases

```python
import re
```

```python
chat1 = "codebasics: you ask lot of questions 1234567891 abcDW@xyz.com"
chat2 = "codebasics: here it is, (123)-456-7891 aX_21c@xyz.com"
chat3 = "codebasics: yes, hone 1234567891 abc@xyz.io"
```

```python
pattern = r'\(?\d{3}\)?-?\d{3}-?\d{4}'

matches = re.findall(pattern, chat2)
matches
```

```python
pattern = r'[a-z0-9A-Z_]*@[a-z0-9A-Z]*\.[a-zA-Z]*'

matches = re.findall(pattern, chat2)
matches
```

```python
chat1 = "order # 12344343443"
chat2 = "order number 12344343443"
chat3 = "order 12344343443"
```

```python
pattern = r"order[^\d]* (\d*)"

matches = re.match(pattern, chat3)
group = matches.group(1)
group
```

```python
def get_pattern_match(pattern, text):
  matches = re.findall(pattern, text)
  if matches:
    return matches[0].strip()

get_pattern_match(pattern, text)
```

```python
def get_personal_information(text):
  age = get_pattern_match(r"age (\d+)", text)
  full_name = get_pattern_match(r"Born(.*)", text)
  birth_date = get_pattern_match(r"Born.*\n(.*)\(age"
, text)
  birth_place = get_pattern_match(r"\(age.*\)\n(.*)", text)
  return {
      "age": age,
      "name": full_name,
      "birth_date": birth_date,
      "birth_place": birth_place
  }
```

```python
get_personal_information(text)
```
