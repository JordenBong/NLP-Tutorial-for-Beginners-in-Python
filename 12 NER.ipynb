# 12 Named Entity Recognition NER

```python
import spacy

nlp = spacy.load("en_core_web_sm")
```

```python
nlp.pipe_names
```

```python
doc = nlp("Tesla Inc is going to acquire Twitter Inc for $45 billion")

for ent in doc.ents:
  print(ent.text, "|", ent.label_, "|", spacy.explain(ent.label_))
```

```python
from spacy import displacy

displacy.render(doc, style="ent")
```

Since there is some mistake made my pretrained ner system in the spacy package, you can actually modify their label manually

Span = Slice of tokens / substring

```python
type(doc[2:5])
```

```python
from spacy.tokens import Span

s1 = Span(doc, 0, 1, label="ORG")
s2 = Span(doc, 5, 6, label="ORG")

doc.set_ents([s1, s2], default="unmodified")
```

```python
for ent in doc.ents:
  print(ent.text, "|", ent.label_, "|", spacy.explain(ent.label_))
```

Extract all the Geographical (cities, Countries, states) names from a given text

```python
text = """Kiran want to know the famous foods in each state of India. So, he opened Google and search for this question. Google showed that
in Delhi it is Chaat, in Gujarat it is Dal Dhokli, in Tamilnadu it is Pongal, in Andhrapradesh it is Biryani, in Assam it is Papaya Khar,
in Bihar it is Litti Chowkha and so on for all other states"""
```

```python
doc = nlp(text)
```

```python
gpes = []

for ent in doc.ents:
  if ent.label_ == "GPE":
    print(ent.text, "|", ent.label_)
    gpes.append(ent)
```

```python
from spacy.tokens import Span

s1 = Span(doc, 0, 1, label="PERSON")

doc.set_ents([s1], default="unmodified")
```

```python
gpes = []

for ent in doc.ents:
  if ent.label_ == "GPE":
    print(ent.text, "|", ent.label_)
    gpes.append(ent)
```
