# Indian Food Recipe - Word Embedding Training - fastText Model Training

```python
import pandas as pd
```

```python
df = pd.read_csv("dataset/Cleaned_Indian_Food_Dataset.csv")
print(df.shape)
df.head()
```

```python
df.TranslatedInstructions[0]
```

```python
import re

text = df.TranslatedInstructions[0]
```

```python
def preprocess(text):
    text = re.sub(r'[^\w\s]', ' ', text)
    text = re.sub(r'[ \n]+', ' ', text)
    return text.strip().lower()
```

```python
preprocess(text)
```

```python
df.TranslatedInstructions = df.TranslatedInstructions.map(preprocess)
```

```python
df.TranslatedInstructions[1]
```

```python
# You must have a specific format file, if you want to print your model
df.to_csv("dataset/food_recipes.txt", columns=["TranslatedInstructions"], header=None, index=False)
```

```python
# train fasttext model 
model = fasttext.train_unsupervised("dataset/food_recipes.txt")
```
